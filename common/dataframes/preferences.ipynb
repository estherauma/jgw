```python
import requests
import pandas as pd
import numpy as np
```

Fetch Preferences data

```python
startDate = "2023-01-01"
endDate = "2023-06-29"

inline_questions = requests.get(f"http://api.justgood.work/api/v2/preferences?startDate={startDate}&endDate={endDate}")
data = inline_questions.json()["data"]
```

Convert data to dataframe

```python
# convert data to pandas dataframe
data_df = pd.DataFrame.from_records(data)

_startDate = "2023-01-01"
_endDate = "2023-06-28"
mask = data_df["createdAt"].between(_startDate, _endDate, inclusive="both")
_df = data_df.loc[mask]

org_stats = _df["organisation"].value_counts()

# save to excel
org_stats
org_stats.to_excel(f'work-codes-{_startDate}-to-{_endDate}.xlsx')
```
