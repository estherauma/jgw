```python
import requests
import pandas as pd
import numpy as np
```

Fetch Preferences data

```python
inline_questions = requests.get("http://api.justgood.work/api/v2/preferences")
data = inline_questions.json()["data"]
```

Convert data to dataframe

```python
preference_df = pd.DataFrame(data)

# view the first 2 records
preference_df.head(1)
```

```python

def filter_by_date(df, start_date, end_date):
    mask = df["createdAt"].between(start_date, end_date, inclusive="both")
    df = df.loc[mask]
    return df

start_date='2023-04-21'
end_date='2023-04-21'
df = filter_by_date(preference_df,start_date,end_date)
df.loc[df["organisation"] == '', "organisation"] = 'Not specified'
df["organisation"].value_counts()
```
