```python
import requests
import pandas as pd
import numpy as np
```

Fetch Preferences data

```python
startDate = "2019-01-01"
endDate = "2023-07-13"

inline_questions = requests.get(f"http://api.justgood.work/api/v2/preferences?startDate={startDate}&endDate={endDate}")
data = inline_questions.json()["data"]
```

Convert data to dataframe

```python
df = pd.DataFrame.from_records(data)
```

```python
def filter_sectors(row):
    search_sector = "construction".lower()
    target_row = row['sectorsOfWork']
    if isinstance(target_row, list):
        return search_sector in [sector.lower() for sector in target_row]
    elif isinstance(target_row, str):
        return target_row.lower() == search_sector
    else:
        return False

filtered_df = df[df.apply(filter_sectors, axis=1)]
filtered_df = filtered_df[filtered_df['location_of_potential_work'].isin(['UK', 'uk', 'GBR', 'United Kingdom'])]
# vv = filtered_df[['location_of_potential_work', 'sectorsOfWork']]
# filtered_df['location_of_potential_work'].value_counts()

# save vv to excel
# filtered_df.to_excel('uk_agriculture.xlsx')

filtered_df
```
