```python
import requests
import pandas as pd
import numpy as np
```

Fetch Preferences data

```python
startDate = "2022-07-01"
endDate = "2023-08-03"

inline_questions = requests.get(f"http://api.justgood.work/api/v2/preferences?startDate={startDate}&endDate={endDate}")
data = inline_questions.json()["data"]
```

Convert data to dataframe

```python
df = pd.DataFrame.from_records(data)
```

```python
# df["route"].value_counts()
#  get a list of the countries names
countries = df["route"].str.split(" -> ", expand=True).stack().unique()
countries
```
