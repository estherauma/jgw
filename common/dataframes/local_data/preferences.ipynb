```python
import pandas as pd
import numpy as np
```

```python
#load excel file
df = pd.read_excel('all_preferences.xlsx')
```

```python
def filter_sectors(row):
    if isinstance(row['sectorsOfWork'], list):
        return 'construction' in [sector.lower() for sector in row['sectorsOfWork']]
    elif isinstance(row['sectorsOfWork'], str):
        return row['sectorsOfWork'].lower() == 'construction'
    else:
        return False

filtered_df = df[df.apply(filter_sectors, axis=1)]
filtered_df = filtered_df[filtered_df['location_of_potential_work'].isin(['UK', 'uk', 'GBR', 'United Kingdom'])]
filtered_df[['location_of_potential_work', 'sectorsOfWork']]
```
