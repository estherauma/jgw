```python
import requests
import pandas as pd
import numpy as np
import polars as pl
```

```python
# Fetch help contacts data
inline_question = requests.get("http://api.justgood.work/api/v2/inlinequestions?startDate=2020-01-01&endDate=2023-06-26")
# inline_question = requests.get("http://api.justgood.work/api/v2/messaging-questions")
inline_question_json = inline_question.json()["data"]
```

```python
# Convert to dataframe
inline_question_df = pd.DataFrame.from_records(inline_question_json)
inline_question_df.head(1)

unwanted_cols = ["", "deviceId", "undefined", "__v", "_id"]
demographic_cols = [
    "appId",
    "locale",
    "route",
    "language",
    "origin",
    "destination",
    "organisation",
    "factoryFarm",
    "createdAt",
    "updatedAt",
]
exclude_cols = [*unwanted_cols, *demographic_cols]
questions_df = inline_question_df[[i for i in inline_question_df.columns if i not in exclude_cols]]
result = {}

for column in questions_df.columns:
    counts = questions_df[column].value_counts()
    print(counts)
    result[column] = {'yes': counts.get('yes', 0), 'no': counts.get('no', 0)}

print(result)
```
